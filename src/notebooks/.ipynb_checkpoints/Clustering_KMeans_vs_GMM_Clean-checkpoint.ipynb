# 📊 So sánh KMeans và Gaussian Mixture Model (GMM)
Pipeline sử dụng Spark MLlib để xử lý dữ liệu, thực hiện phân cụm bằng cả KMeans và GMM,
đồng thời trực quan hóa kết quả bằng PCA và tính toán Silhouette Score để đánh giá chất lượng phân cụm.

```python
from pyspark.ml.feature import StandardScaler, PCA
from pyspark.ml.clustering import KMeans, GaussianMixture
from pyspark.ml.evaluation import ClusteringEvaluator
from pyspark.sql.functions import lit
import matplotlib.pyplot as plt

# ---------------- PREPROCESSING ---------------- #
def scale_features(df, input_col="features", output_col="scaled_features"):
    scaler = StandardScaler(inputCol=input_col, outputCol=output_col, withStd=True, withMean=False)
    return scaler.fit(df).transform(df)

def apply_pca(df, k=2, input_col="scaled_features", output_col="pca_features"):
    pca = PCA(k=k, inputCol=input_col, outputCol=output_col)
    return pca.fit(df).transform(df)

def pca_to_pandas(df, pca_col="pca_features", cluster_col="cluster", k_col="k_value"):
    return df.select(pca_col, cluster_col, k_col).rdd.map(
        lambda row: (float(row[pca_col][0]), float(row[pca_col][1]),
                     row[cluster_col], row[k_col])
    ).toDF(["x", "y", "cluster", "k_value"]).toPandas()
```

```python
# ---------------- CLUSTERING CORE ---------------- #
def run_clustering(algorithm, k_values, scaled_df, features_col="scaled_features", prediction_col="cluster", label="KMeans"):
    results, clustered_dfs = {}, {}
    
    for k in k_values:
        print(f"\n== {label} với k = {k} ==")
        model = (
            KMeans(k=k, seed=42, featuresCol=features_col, predictionCol=prediction_col)
            if algorithm == "kmeans"
            else GaussianMixture(k=k, seed=42, featuresCol=features_col, predictionCol=prediction_col)
        )
        model_fit = model.fit(scaled_df)
        clustered_df = model_fit.transform(scaled_df).withColumn("k_value", lit(k))
        clustered_dfs[k] = clustered_df

        evaluator = ClusteringEvaluator(
            featuresCol=features_col,
            predictionCol=prediction_col,
            metricName="silhouette",
            distanceMeasure="squaredEuclidean"
        )
        score = evaluator.evaluate(clustered_df)
        print(f"Silhouette Score ({label}): {score:.4f}")
        results[k] = score

    return results, clustered_dfs
```

```python
# ---------------- VISUALIZATION ---------------- #
def visualize_clusters(clustered_dfs, label="KMeans"):
    combined_df = None
    for k, df in clustered_dfs.items():
        if combined_df is None:
            combined_df = df
        else:
            combined_df = combined_df.union(df)
    
    pca_df = apply_pca(combined_df)
    pandas_data = pca_to_pandas(pca_df, cluster_col="cluster", k_col="k_value")

    unique_k = sorted(clustered_dfs.keys())
    fig, axes = plt.subplots(1, len(unique_k), figsize=(6*len(unique_k), 5))
    axes = axes if len(unique_k) > 1 else [axes]

    for i, k in enumerate(unique_k):
        data = pandas_data[pandas_data["k_value"] == k]
        for cluster_id in sorted(data["cluster"].unique()):
            cluster_data = data[data["cluster"] == cluster_id]
            axes[i].scatter(cluster_data["x"], cluster_data["y"], label=f"Cụm {cluster_id}", s=50, alpha=0.6)
        axes[i].set_title(f"{label} với k={k}")
        axes[i].legend()
        axes[i].grid(True)

    plt.suptitle(f"Trực quan hóa phân cụm bằng {label}", fontsize=14)
    plt.show()
```

```python
def show_silhouette_scores(results, label="KMeans"):
    print(f"\n== Tổng kết Silhouette Score - {label} ==")
    print("k\tScore")
    best_k = max(results, key=results.get)
    for k, score in sorted(results.items()):
        print(f"{'→' if k == best_k else ' '} {k}\t{score:.4f}")

    plt.figure(figsize=(8, 5))
    plt.bar(results.keys(), results.values(), color="skyblue")
    plt.axvline(best_k, color="red", linestyle="--", label=f"Tốt nhất: k={best_k}")
    plt.title(f"Silhouette Score - {label}")
    plt.xlabel("k")
    plt.ylabel("Score")
    plt.legend()
    plt.grid(True)
    plt.show()
```

```python
# ---------------- PIPELINE EXECUTION ---------------- #
def run_pipeline(original_df, k_values=[3, 4, 5]):
    scaled_df = scale_features(original_df)

    # KMeans
    kmeans_results, kmeans_dfs = run_clustering("kmeans", k_values, scaled_df, label="KMeans")
    visualize_clusters(kmeans_dfs, label="KMeans")
    show_silhouette_scores(kmeans_results, label="KMeans")

    # GMM
    gmm_results, gmm_dfs = run_clustering("gmm", k_values, scaled_df, label="GMM")
    visualize_clusters(gmm_dfs, label="GMM")
    show_silhouette_scores(gmm_results, label="GMM")

    return kmeans_results, gmm_results
```
